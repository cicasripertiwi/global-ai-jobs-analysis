```python
import pandas as pd
df = pd.read_csv('/content/global_ai_jobs.csv')
average_salary_usd = df.groupby('job_role')['salary_usd'].mean().sort_values(ascending=False)
print(average_salary_usd)
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import matplotlib.pyplot as plt

average_salary_usd.plot(kind='bar')

plt.title('Average Salary in USD by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Average Salary (USD)')
plt.xticks(rotation=45)
plt.show()
```

Based on the chart, Research Scientist has the highest average salary, at approximately USD 109,797, while Data Analyst has the lowest average salary, at approximately USD 69,429.

The other job roles, such as Machine Learning Engineer, Computer Vision Engineer, NLP Engineer, AI Engineer, Software Engineer AI, and Data Scientist, have average salaries between these two positions.

This show that, in this dataset, roles related to AI, machine learning, and research tend to have higher average salaries than data analysis roles. However, salary can also be influenced by other factors, such as experience level, country, company size, and work arrangement.

Overall, Research Scientist has the highest average salary, while Data Analyst has the lowest average salary among the job roles analyzed.

```python
job_openings_country = df.groupby('country')['job_openings'].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

job_openings_country.plot(kind='bar')

plt.title('Job Openings by Country')
plt.xlabel('Country')
plt.ylabel('Job Openings')
plt.xticks(rotation=45)
plt.show()
```

Based on the chart, Singapore has the highest number of job openings at approximately 134,000, followed by Australia and Canada. Meanwhile, Japan has the lowest number of job openings at around 129,000.

The results show that job opportunities in the AI and Data Science industry are available across all countries included in the dataset. Although Singapore has the highest number of openings, the differences between countries are relatively small.

This suggests that the demand for AI and Data Science professionals is distributed across multiple countries. However, the number of job openings alone does not indicate how easy it is to get a job, as competition, required skills, and hiring requirements may also vary.

Overall, this analysis provides an overview of job opportunities across different countries in the global AI and Data Science market.
